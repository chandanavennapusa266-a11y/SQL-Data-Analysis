# Data Analytics Project 3 — SQL Data Analysis

This notebook demonstrates the SQL analysis required for Project 3 using a SQLite database.

**Core SQL skills:** SELECT, WHERE, ORDER BY, GROUP BY, COUNT, SUM, AVG, HAVING.


```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("ecommerce_analysis.db")
```

## 1. SELECT — View the data

```python
query = '''
SELECT *
FROM orders
LIMIT 20;
'''
display(pd.read_sql_query(query, conn))
```

## 2. WHERE — Filter rows

```python
query = '''
SELECT OrderID, CustomerID, Product, Quantity, TotalPrice
FROM orders
WHERE Quantity > 2
ORDER BY Quantity DESC;
'''
display(pd.read_sql_query(query, conn))
```

## 3. ORDER BY — Sort results

```python
query = '''
SELECT OrderID, CustomerID, Product, TotalPrice
FROM orders
ORDER BY TotalPrice DESC
LIMIT 20;
'''
display(pd.read_sql_query(query, conn))
```

## 4. GROUP BY + COUNT + SUM + AVG — Product performance

```python
query = '''
SELECT
    Product,
    COUNT(OrderID) AS order_count,
    SUM(TotalPrice) AS total_revenue,
    AVG(TotalPrice) AS average_order_value
FROM orders
GROUP BY Product
ORDER BY total_revenue DESC;
'''
display(pd.read_sql_query(query, conn))
```

## 5. GROUP BY — Payment methods

```python
query = '''
SELECT
    PaymentMethod,
    COUNT(OrderID) AS order_count,
    SUM(TotalPrice) AS total_revenue,
    AVG(TotalPrice) AS average_order_value
FROM orders
GROUP BY PaymentMethod
ORDER BY total_revenue DESC;
'''
display(pd.read_sql_query(query, conn))
```

## 6. HAVING — Filter grouped results

```python
query = '''
SELECT
    CustomerID,
    COUNT(OrderID) AS order_count,
    SUM(TotalPrice) AS total_spent
FROM orders
GROUP BY CustomerID
HAVING COUNT(OrderID) >= 2
   AND SUM(TotalPrice) > 1000
ORDER BY total_spent DESC;
'''
display(pd.read_sql_query(query, conn))
```

## 7. Monthly analysis

```python
query = '''
SELECT
    strftime('%Y-%m', Date) AS order_month,
    COUNT(OrderID) AS order_count,
    SUM(TotalPrice) AS total_revenue,
    AVG(TotalPrice) AS average_order_value
FROM orders
GROUP BY strftime('%Y-%m', Date)
ORDER BY order_month;
'''
display(pd.read_sql_query(query, conn))
```

## 8. Percentage contribution

```python
query = '''
SELECT
    Product,
    SUM(TotalPrice) AS product_revenue,
    ROUND(
        100.0 * SUM(TotalPrice) / (SELECT SUM(TotalPrice) FROM orders),
        2
    ) AS revenue_percentage
FROM orders
GROUP BY Product
ORDER BY revenue_percentage DESC;
'''
display(pd.read_sql_query(query, conn))
```

## 9. Data-quality check

```python
query = '''
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT OrderID) AS unique_order_ids,
    COUNT(*) - COUNT(DISTINCT OrderID) AS duplicate_order_id_count
FROM orders;
'''
display(pd.read_sql_query(query, conn))
```
